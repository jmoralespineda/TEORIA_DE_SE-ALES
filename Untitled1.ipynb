Analizar la semejanza de las distribuciones resultantes del espectro promediado, con varias realizaciones, respecto a una gaussiana,
comparando la desviación de los armónicos contra diferentes valores de desviación del ruido para las varias realizaciones.

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, shapiro
import seaborn as sns

# --------------------------------------
# Parámetros globales
# --------------------------------------
fs = 1000     # Frecuencia de muestreo (Hz)
T = 1024      # Número de muestras por realización
M = 200       # Número de realizaciones
A = 1.0       # Amplitud de la señal
f0 = 250      # Frecuencia de la señal (Hz)
device = "cuda" if torch.cuda.is_available() else "cpu"

# --------------------------------------
# Funciones auxiliares
# --------------------------------------

def generar_senales(sigma, M=M, T=T, fs=fs, A=A, f0=f0, device=device):
    """Genera M realizaciones de una señal cosenoidal + ruido gaussiano."""
    t = torch.arange(T, device=device) / fs
    s = A * torch.cos(2 * torch.pi * f0 * t)
    s = s.expand(M, -1)  # replicar para todas las realizaciones
    noise = torch.normal(0.0, sigma, size=(M, T), device=device)
    x = s + noise
    return x

def periodograma(x, fs=fs):
    """Calcula periodograma de un batch de señales (M,T)."""
    X = torch.fft.fft(x, dim=1)
    Sxx = (torch.abs(X) ** 2) / x.shape[1]
    freqs = torch.fft.fftfreq(x.shape[1], 1/fs)
    return freqs.cpu().numpy(), Sxx.cpu().numpy()

def analizar_armonico(Sxx, freqs, f0=f0):
    """Extrae las amplitudes espectrales en la frecuencia más cercana a f0."""
    idx = np.argmin(np.abs(freqs - f0))
    valores = Sxx[:, idx]  # un valor por realización
    return valores

def comparar_con_gauss(valores, sigma_ruido, ax):
    """Grafica histograma de valores y superpone una gaussiana."""
    mu, sigma = np.mean(valores), np.std(valores)

    # Histograma
    sns.histplot(valores, bins=20, stat="density", alpha=0.5, ax=ax, label="Datos")

    # Gaussiana de referencia
    x = np.linspace(min(valores), max(valores), 200)
    y = norm.pdf(x, mu, sigma)
    ax.plot(x, y, 'r--', label=f"N({mu:.2f},{sigma:.2f}²)")

    # Test de Shapiro-Wilk (normalidad)
    stat, pval = shapiro(valores)
    resultado = "Gaussiana" if pval > 0.05 else "No Gaussiana"

    ax.set_title(f"σ ruido = {sigma_ruido} → {resultado}")
    ax.legend()
    ax.grid(True)

# --------------------------------------
# Main
# --------------------------------------
def main():
    # Valores de desviación del ruido a probar
    sigma_vals = [0.5, 1.0, 2.0, 3.5, 5.0, 10.0]

    fig, axes = plt.subplots(2, 3, figsize=(16, 8))
    axes = axes.ravel()

    for i, sigma_ruido in enumerate(sigma_vals):
        # 1. Generar señales
        x = generar_senales(sigma_ruido)

        # 2. Periodograma por realización
        freqs, Sxx = periodograma(x)

        # 3. Extraer valores espectrales en f0
        valores = analizar_armonico(Sxx, freqs)

        # 4. Comparar distribución con gaussiana
        comparar_con_gauss(valores, sigma_ruido, axes[i])

    plt.suptitle("Distribución de la potencia espectral en f0 para distintas σ del ruido")
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    main()
```

#También deben incluir la estimación de la PSD con los métodos de Periodograma, Welch y Multitaper, para este caso puede ser una señal con más de un armónico.


```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm, shapiro
from scipy.signal.windows import dpss

# ============================================================
# 1. Generación de señales
# ============================================================

def generar_senal_armonicos(T, fs, M, A_list, f_list, sigma, device="cpu"):
    """
    Genera M realizaciones de una señal compuesta por varios armónicos + ruido gaussiano.
    A_list: lista de amplitudes
    f_list: lista de frecuencias correspondientes
    sigma : desviación estándar del ruido
    """
    t = torch.arange(T, device=device) / fs
    s = torch.zeros(T, device=device)
    for A, f0 in zip(A_list, f_list):
        s += A * torch.cos(2 * torch.pi * f0 * t)
    s = s.expand(M, -1)  # replicar señal determinista
    noise = torch.normal(0.0, sigma, size=(M, T), device=device)
    x = s + noise
    return x

# ============================================================
# 2. Estimadores de PSD
# ============================================================

def periodograma(x, fs):
    """Estimación de PSD por periodograma clásico (por realización)."""
    N = x.shape[1]
    X = torch.fft.fft(x, dim=1)
    Sxx = (torch.abs(X) ** 2) / N
    freqs = torch.fft.fftfreq(N, 1/fs)
    return freqs.cpu().numpy(), Sxx.cpu().numpy()

def welch(x, fs, L, overlap=0.5, window="hamming"):
    """
    Estimación de PSD con método de Welch (implementación manual).
    - x: batch (M,T)
    - L: longitud del segmento
    - overlap: fracción de solapamiento
    - window: tipo de ventana (solo 'hamming' implementado)
    """
    M, N = x.shape
    step = int(L * (1 - overlap))
    num_segments = (N - L) // step + 1

    if window == "hamming":
        w = torch.hamming_window(L, periodic=False, device=x.device)
    else:
        w = torch.ones(L, device=x.device)

    U = (w**2).sum() / L  # normalización

    psd_list = []
    for start in range(0, N-L+1, step):
        seg = x[:, start:start+L] * w
        X = torch.fft.fft(seg, dim=1)
        P = (torch.abs(X) ** 2) / (L * U)
        psd_list.append(P)

    Sxx = torch.stack(psd_list).mean(dim=0)  # promedio sobre segmentos
    freqs = torch.fft.fftfreq(L, 1/fs)
    return freqs.cpu().numpy(), Sxx.cpu().numpy()

def multitaper(x, fs, NW=3, K=5):
    """
    Estimación de PSD multitaper (Thomson).
    - NW: time-bandwidth product
    - K : número de tapers
    """
    M, N = x.shape
    tapers, eigs = dpss(N, NW, K, return_ratios=True)
    # Make a copy to avoid negative strides issue with torch.tensor
    tapers = torch.tensor(tapers.copy(), dtype=torch.float32, device=x.device)

    psd_list = []
    for k in range(K):
        seg = x * tapers[k]
        X = torch.fft.fft(seg, dim=1)
        P = (torch.abs(X) ** 2) / N
        psd_list.append(P)

    Sxx = torch.stack(psd_list).mean(dim=0)  # promedio sobre tapers
    freqs = torch.fft.fftfreq(N, 1/fs)
    return freqs.cpu().numpy(), Sxx.cpu().numpy()

# ============================================================
# 3. Análisis estadístico de armónicos
# ============================================================

def analizar_armonico(Sxx, freqs, f0):
    """Extrae los valores de potencia espectral en la frecuencia más cercana a f0."""
    idx = np.argmin(np.abs(freqs - f0))
    valores = Sxx[:, idx]
    return valores

def comparar_con_gauss(valores, sigma_ruido, f0, ax):
    """Grafica histograma + gaussiana y test de normalidad."""
    mu, sigma = np.mean(valores), np.std(valores)
    sns.histplot(valores, bins=20, stat="density", alpha=0.5, ax=ax, label="Datos")
    x = np.linspace(min(valores), max(valores), 200)
    y = norm.pdf(x, mu, sigma)
    ax.plot(x, y, 'r--', label=f"N({mu:.2f},{sigma:.2f}²)")
    stat, pval = shapiro(valores)
    resultado = "≈ Gaussiana" if pval > 0.05 else "No Gaussiana"
    ax.set_title(f"f0={f0} Hz, σ={sigma_ruido} → {resultado}")
    ax.legend()
    ax.grid(True)

# ============================================================
# 4. Main
# ============================================================

def main():
    fs, T, M = 1000, 2048, 200
    A_list = [1.0, 0.8]   # amplitudes
    f_list = [250, 400]   # frecuencias (Hz)
    sigma_vals = [0.5, 2.0, 5.0, 10.0]
    device = "cuda" if torch.cuda.is_available() else "cpu"

    for sigma in sigma_vals:
        # --------------------------
        # Generación
        # --------------------------
        x = generar_senal_armonicos(T, fs, M, A_list, f_list, sigma, device=device)

        # --------------------------
        # Estimación de PSD
        # --------------------------
        freqs_p, Sxx_p = periodograma(x, fs)
        freqs_w, Sxx_w = welch(x, fs, L=512, overlap=0.5)
        freqs_mt, Sxx_mt = multitaper(x, fs, NW=3, K=5)

        # Promedios (para graficar espectro)
        PSD_p = Sxx_p.mean(axis=0)
        PSD_w = Sxx_w.mean(axis=0)
        PSD_mt = Sxx_mt.mean(axis=0)

        # --------------------------
        # Graficar PSD
        # --------------------------
        plt.figure(figsize=(14,5))
        plt.plot(freqs_p, 10*np.log10(PSD_p+1e-12), label="Periodograma")
        plt.plot(freqs_w, 10*np.log10(PSD_w+1e-12), label="Welch")
        plt.plot(freqs_mt, 10*np.log10(PSD_mt+1e-12), label="Multitaper")
        for f0 in f_list:
            plt.axvline(f0, color='r', linestyle='--', label=f"{f0} Hz")
        plt.title(f"Estimación de PSD para σ={sigma}")
        plt.xlabel("Frecuencia [Hz]")
        plt.ylabel("PSD [dB]")
        plt.legend()
        plt.grid(True)
        plt.show()

        # --------------------------
        # Distribución en armónicos
        # --------------------------
        fig, axes = plt.subplots(1, len(f_list), figsize=(12,4))
        if len(f_list) == 1:
            axes = [axes]
        for i, f0 in enumerate(f_list):
            valores = analizar_armonico(Sxx_p, freqs_p, f0)  # usar Periodograma
            comparar_con_gauss(valores, sigma, f0, axes[i])
        plt.suptitle(f"Distribución de potencias espectrales en armónicos (σ={sigma})")
        plt.tight_layout()
        plt.show()

if __name__ == "__main__":
    main()
```

# **Comparacion con Python**

```python
import torch
import matplotlib.pyplot as plt
import numpy as np
import time

# -----------------------------
# Parámetros de la simulación
# -----------------------------
fs = 1000          # Frecuencia de muestreo (Hz)
T = 1024           # Número de muestras por realización
M = 200            # Número de realizaciones
sigma = 3.5        # Desviación estándar del ruido
A = 1.0            # Amplitud del armónico
f0 = 250           # Frecuencia del armónico (Hz)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Usando: {device}")

# -----------------------------
# Ventana de Hamming
# -----------------------------
hamming_window = torch.hamming_window(T, periodic=False, device=device)

# -----------------------------
# Timer start
# -----------------------------
start_time = time.time()

# -----------------------------
# Generación de todas las señales en paralelo
# -----------------------------
t = torch.arange(T, device=device) / fs
s = A * torch.cos(2 * torch.pi * f0 * t)          # señal determinista
s = s.expand(M, -1)                               # replicar para todas las realizaciones
noise = torch.normal(0.0, sigma, size=(M,T), device=device)
x = s + noise                                     # señales con ruido

# Con ventana de Hamming (broadcasting)
x_hamming = x * hamming_window

# -----------------------------
# Autocorrelación vía FFT (Wiener–Khinchin)
# -----------------------------
def autocorr_fft(batch_signals):
    # FFT de cada señal
    X = torch.fft.fft(batch_signals, n=2*T-1, dim=1)
    Sxx = torch.abs(X) ** 2
    Rxx = torch.fft.ifft(Sxx, dim=1).real
    return Rxx

R_no_window = autocorr_fft(x)
R_hamming   = autocorr_fft(x_hamming)

# Promediar sobre realizaciones
R_avg_no_window = R_no_window.mean(dim=0)
R_avg_hamming   = R_hamming.mean(dim=0)

# Quitar componente DC
R_avg_no_window -= R_avg_no_window.mean()
R_avg_hamming   -= R_avg_hamming.mean()

# -----------------------------
# Timer end
# -----------------------------
end_time = time.time()
print(f"Tiempo total para {M} realizaciones: {end_time - start_time:.4f} segundos")

# -----------------------------
# PSD via Wiener–Khinchin
# -----------------------------
def compute_psd(R_avg):
    Sx = torch.fft.fftshift(torch.fft.fft(torch.fft.ifftshift(R_avg)))
    freqs = torch.fft.fftshift(torch.fft.fftfreq(len(R_avg), d=1/fs))
    PSD = torch.abs(Sx) / len(R_avg)
    return freqs, PSD

freqs_no_window, PSD_no_window = compute_psd(R_avg_no_window)
freqs_hamming,   PSD_hamming   = compute_psd(R_avg_hamming)

# -----------------------------
# Pasar a NumPy para graficar
# -----------------------------
freqs_no_window_np = freqs_no_window.cpu().numpy()
freqs_hamming_np   = freqs_hamming.cpu().numpy()
PSD_no_window_np   = PSD_no_window.cpu().numpy()
PSD_hamming_np     = PSD_hamming.cpu().numpy()

# -----------------------------
# Comparación de PSDs
# -----------------------------
plt.figure(figsize=(14,5))
plt.plot(freqs_no_window_np, 10*np.log10(PSD_no_window_np + 1e-12), label="Sin ventana")
plt.plot(freqs_hamming_np, 10*np.log10(PSD_hamming_np + 1e-12), label="Con ventana Hamming")
plt.axvline(f0, color='r', linestyle='--', label=f"f0 = {f0} Hz")
plt.xlabel("Frecuencia [Hz]")
plt.ylabel("PSD [dB]")
plt.title("Comparación de estimación de PSD (Wiener–Khinchin) con y sin ventaneo")
plt.grid(True)
plt.legend()
plt.ylim(-60, None)
plt.show()
```

# **Similitudes a una Gaussiana**

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import shapiro

# Set a random seed for reproducibility
torch.manual_seed(42)

# --- 1. Simulation Parameters ---
fs = 1000        # Sampling frequency (Hz)
T = 1024         # Number of samples per realization
M = 200          # Number of realizations for one estimation of 'a'
K = 300           # Number of times to repeat the estimation process
A = 1.0          # True amplitude of the harmonic
f0 = 250         # Harmonic frequency (Hz)
phi = 0.0        # Harmonic phase

# Noise sigma values to test
sigma_values = torch.linspace(0.1, 30, 20)
# Y-axis grid lines
grid_y = [1, 3, 5, 7, 10, 15, 17, 20, 27]

# --- 2. Lists to store the results ---
mean_a_averages = []
is_gaussian = []

# --- 3. Main Loop over sigma values ---
for sigma in sigma_values:
    a_averages = torch.zeros(K)

    # Create a time vector
    t = torch.arange(T, dtype=torch.float32) / fs
    s = A * torch.cos(2 * torch.pi * f0 * t + phi)

    for k in range(K):
        # Vectorized generation of M noisy signals
        noise = torch.normal(0, sigma, size=(M, T))
        x = s + noise

        # Vectorized estimation of 'a' for all M signals
        a_estimated = torch.max(torch.abs(x), dim=1).values

        # Calculate the average for this run and store it
        a_averages[k] = torch.mean(a_estimated)

    # Store the mean of the K averages
    mean_of_averages = torch.mean(a_averages).item()
    mean_a_averages.append(mean_of_averages)

    # Perform Gaussianity test
    # Convert PyTorch tensor to a NumPy array for the Shapiro-Wilk test
    shapiro_stat, p_value = shapiro(a_averages.numpy())

    # Check if p-value is greater than 0.05
    is_gaussian.append(p_value > 0.05)

# --- 4. Graph Construction ---
colors = ['green' if g else 'red' for g in is_gaussian]

plt.figure(figsize=(12, 8))
plt.scatter(sigma_values.numpy(), mean_a_averages, c=colors, s=100)
plt.plot(sigma_values.numpy(), mean_a_averages, c='gray', linestyle='--')
plt.xlabel("Noise standard deviation (sigma)")
plt.ylabel("Mean of estimated amplitude (Mean of a)")
plt.title("Mean of Estimated Amplitude vs. Noise Standard Deviation")
plt.grid(True, which='both', axis='y', linestyle='--', linewidth=0.5, color='gray')

# Add custom grid lines
for y_val in grid_y:
    plt.axhline(y=y_val, color='gray', linestyle='--', linewidth=0.5)

# Create a custom legend for colors
plt.plot([], [], 'o', color='green', label='Is Gaussian (p > 0.05)')
plt.plot([], [], 'o', color='red', label='Is Not Gaussian (p ≤ 0.05)')
plt.legend()

plt.show()
```

# **Señal con ruido - PROMEDIO DE AUTOCORRELACION - PSD**

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Parámetros de la simulación
# -----------------------------
fs = 1000          # Frecuencia de muestreo (Hz)
T = 1024           # Número de muestras por realización
M = 200            # Número de realizaciones
sigma = 3.5        # Desviación estándar del ruido
A = 1.0            # Amplitud del armónico
f0 = 250            # Frecuencia del armónico (Hz)

# -----------------------------
# Inicializar acumulador de autocorrelaciones
# -----------------------------
R_sum = np.zeros(2*T-1)

# -----------------------------
# Bucle sobre realizaciones
# -----------------------------
for m in range(M):
    # Vector de tiempo
    t = np.arange(T) / fs

    # Ruido blanco gaussiano
    noise = np.random.normal(0, sigma, T)

    # Armónico
    s = A * np.cos(2 * np.pi * f0 * t)

    # Señal total
    x = s + noise

    # Autocorrelación (modo "full")
    r = np.correlate(x, x, mode='full')

    # Normalización "unbiased" (divide por nº de muestras válidas en cada lag)
    lags = np.arange(-T+1, T)
    r = r / (T - np.abs(lags))

    # Acumular
    R_sum += r

# -----------------------------
# Autocorrelación promedio
# -----------------------------
R_avg = R_sum / M
lags = np.arange(-T+1, T) / fs  # eje de tiempo para los lags

# -----------------------------
# PSD via Wiener–Khinchin
# -----------------------------
Sx = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(R_avg)))
freqs = np.fft.fftshift(np.fft.fftfreq(len(R_avg), d=1/fs))
PSD = np.abs(Sx)

# -----------------------------
# Gráficas
# -----------------------------

plt.figure(figsize=(14,4))
plt.plot(t, x, label="Última realización")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")
plt.title("Señal (Armónico + Ruido blanco gaussiano)")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14,4))
plt.plot(lags, R_avg)
plt.xlabel("Lag [s]")
plt.ylabel("Rxx(lag)")
plt.title("Autocorrelación promedio (M realizaciones)")
plt.grid(True)
plt.show()

plt.figure(figsize=(14,4))
plt.plot(freqs, 10*np.log10(PSD + 1e-12))  # escala dB, evitar log(0)
plt.xlabel("Frecuencia [Hz]")
plt.ylabel("PSD [dB]")
plt.title("Densidad espectral de potencia (PSD) vía Wiener–Khinchin")
plt.grid(True)
plt.ylim(-30, 30)
plt.show()
```

# **Histograma de Amplitudes**

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# Set seed for reproducibility
torch.manual_seed(42)

# -----------------------------
# Simulation Parameters
# -----------------------------
fs = 1000        # Sampling frequency (Hz)
T = 1024         # Number of samples per realization
M = 200          # Number of realizations
sigma = 3.5      # Standard deviation of the noise
A = 1.0          # Harmonic amplitude (The value we want to estimate)
f0 = 250         # Harmonic frequency (Hz)

# -----------------------------
# Initialize amplitude accumulator
# -----------------------------
a_sum = 0.0
# Array to store each amplitude estimate
a_estimates = []

# -----------------------------
# Loop over realizations
# -----------------------------
for m in range(M):
    # Time vector
    t = torch.arange(T, dtype=torch.float32) / fs

    # Gaussian white noise
    noise = torch.normal(0, sigma, size=(T,))

    # Harmonic
    s = A * torch.cos(2 * torch.pi * f0 * t)

    # Total signal
    x = s + noise

    # Estimate amplitude 'a' from this realization
    # We use the maximum absolute value of the signal
    a_estimated = torch.max(torch.abs(x)).item()

    # Sum the estimate and store it
    a_sum += a_estimated
    a_estimates.append(a_estimated)

# -----------------------------
# Calculate average amplitude
# -----------------------------
a_average = a_sum / M

print(f"Número de realizaciones (M): {M}")
print(f"Amplitud verdadera (A): {A}")
print(f"Amplitud promedio estimada (a_average): {a_average:.4f}")

# Convert list to numpy array for histogram plotting
a_estimates_np = np.array(a_estimates)

# -----------------------------
# Plotting the histogram
# -----------------------------
plt.figure(figsize=(10, 6))
plt.hist(a_estimates_np, bins=30, density=True, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(a_average, color='red', linestyle='--', linewidth=2, label=f'Amplitud Promedio: {a_average:.4f}')
plt.axvline(A, color='green', linestyle='-', linewidth=2, label=f'Amplitud Verdadera: {A}')
plt.title('Histograma de Amplitudes Estimadas', fontsize=16)
plt.xlabel('Amplitud Estimada', fontsize=12)
plt.ylabel('Densidad de Frecuencia', fontsize=12)
plt.legend()
plt.grid(axis='y', linestyle='--')
plt.show()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# Establecer la semilla para reproducibilidad
torch.manual_seed(42)

# --- 1. Parámetros de la Simulación ---
fs = 1000        # Frecuencia de muestreo (Hz)
T = 1024         # Número de muestras por realización
M = 200          # Número de realizaciones
sigma = 3.5      # Desviación estándar del ruido
A = 1.0          # Amplitud del armónico (el valor a estimar)
f0 = 250         # Frecuencia del armónico (Hz)

# --- 2. Inicialización de Acumuladores ---
# Arreglo para almacenar cada estimación individual de 'a'
a_estimates = []

# --- 3. Bucle de Estimación ---
for m in range(M):
    # Vector de tiempo
    t = torch.arange(T, dtype=torch.float32) / fs

    # Ruido blanco gaussiano
    noise = torch.normal(0, sigma, size=(T,))

    # Armónico
    s = A * torch.cos(2 * torch.pi * f0 * t)

    # Señal total
    x = s + noise

    # Estimación de la amplitud 'a' a partir de esta realización
    a_estimated = torch.max(torch.abs(x)).item()

    # Almacenar la estimación
    a_estimates.append(a_estimated)

# Convertir la lista a un tensor de PyTorch para los cálculos estadísticos
a_estimates_tensor = torch.tensor(a_estimates)

# --- 4. Cálculo de Estadísticas ---
# Media de las estimaciones
a_mean = torch.mean(a_estimates_tensor)
# Varianza de las estimaciones
a_variance = torch.var(a_estimates_tensor)

print(f"Número de realizaciones (M): {M}")
print(f"Amplitud verdadera (A): {A}")
print(f"Media de las estimaciones de amplitud: {a_mean:.4f}")
print(f"Varianza de las estimaciones de amplitud: {a_variance:.4f}")

# --- 5. Gráfico del Histograma ---
a_estimates_np = a_estimates_tensor.numpy()

plt.figure(figsize=(10, 6))
plt.hist(a_estimates_np, bins=30, density=True, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(a_mean, color='red', linestyle='--', linewidth=2, label=f'Media: {a_mean:.4f}')
plt.axvline(A, color='green', linestyle='-', linewidth=2, label=f'Valor Verdadero: {A}')
plt.title('Distribución de las Estimaciones de Amplitud', fontsize=16)
plt.xlabel('Amplitud Estimada', fontsize=12)
plt.ylabel('Densidad de Frecuencia', fontsize=12)
plt.legend()
plt.grid(axis='y', linestyle='--')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import welch, correlate

# Aplicar un estilo de Matplotlib para un mejor aspecto
plt.style.use('seaborn-v0_8-darkgrid')

# --- 1. Parámetros de Generación de Señales ---
sampling_freq = 8192  # Hz
duration = 2  # segundos
num_samples = int(sampling_freq * duration)
time = np.linspace(0, duration, num_samples, endpoint=False)
frequencies = [10, 100, 1000]  # Frecuencias de las ondas sinusoidales en Hz
noise_std = 0.5  # Desviación estándar del ruido
num_averages = 100  # Número de realizaciones de la señal para el promediado

# --- Estructura para almacenar los resultados ---
stats_results = {
    'freq': [],
    'mean_single': [], 'var_single': [],
    'mean_avg': [], 'var_avg': []
}

# --- 2. Preparación de Gráficos ---
fig_psd, axes_psd = plt.subplots(3, 1, figsize=(12, 12), sharex=True)
fig_autocorr, axes_autocorr = plt.subplots(3, 1, figsize=(12, 12), sharex=True)
fig_hist, axes_hist = plt.subplots(3, 1, figsize=(12, 12), sharex=True)

fig_psd.suptitle('Comparación del Espectro de Potencia', fontsize=16, fontweight='bold')
fig_autocorr.suptitle('Comparación de la Autocorrelación', fontsize=16, fontweight='bold')
fig_hist.suptitle('Comparación del Histograma de Amplitud', fontsize=16, fontweight='bold')

# --- 3. Bucle Principal de Procesamiento ---
for i, freq in enumerate(frequencies):
    # Genera una única señal con ruido como referencia
    signal_single_noisy = np.sin(2 * np.pi * freq * time) + np.random.normal(0, noise_std, num_samples)

    # Genera las señales para el promediado
    signals_to_average = np.zeros(num_samples)
    for _ in range(num_averages):
        noise = np.random.normal(0, noise_std, num_samples)
        signal = np.sin(2 * np.pi * freq * time) + noise
        signals_to_average += signal

    # Calcula la señal promediada
    signal_averaged = signals_to_average / num_averages

    # --- Cálculo de Media y Varianza ---
    mean_single = np.mean(signal_single_noisy)
    var_single = np.var(signal_single_noisy)
    mean_avg = np.mean(signal_averaged)
    var_avg = np.var(signal_averaged)

    # Almacena los resultados
    stats_results['freq'].append(freq)
    stats_results['mean_single'].append(mean_single)
    stats_results['var_single'].append(var_single)
    stats_results['mean_avg'].append(mean_avg)
    stats_results['var_avg'].append(var_avg)

    # --- Espectro de Potencia ---
    f_single, pxx_single = welch(signal_single_noisy, sampling_freq, nperseg=1024)
    f_avg, pxx_avg = welch(signal_averaged, sampling_freq, nperseg=1024)

    # Colores cambiados a rojo y verde
    axes_psd[i].semilogy(f_single, pxx_single, label='Señal con Ruido', alpha=0.7, color='#E41A1C', linestyle='--') # Rojo
    axes_psd[i].semilogy(f_avg, pxx_avg, label='Señal Promediada', linewidth=2, color='#4DAF4A') # Verde
    axes_psd[i].set_title(f'Frecuencia: {freq} Hz', fontsize=14)
    axes_psd[i].set_xlabel('Frecuencia (Hz)', fontsize=12)
    axes_psd[i].set_ylabel('Potencia/Hz (log)', fontsize=12)
    axes_psd[i].set_xlim(0, max(frequencies) + 500)
    axes_psd[i].legend(loc='upper right')

    # --- Autocorrelación ---
    # Normalización para que el valor en el retardo cero sea 1
    autocorr_single = correlate(signal_single_noisy, signal_single_noisy, mode='same')
    autocorr_single /= np.var(signal_single_noisy) * num_samples

    autocorr_avg = correlate(signal_averaged, signal_averaged, mode='same')
    autocorr_avg /= np.var(signal_averaged) * num_samples

    lags_time = np.arange(-num_samples // 2, num_samples // 2) / sampling_freq

    # Colores cambiados a rojo y verde
    axes_autocorr[i].plot(lags_time, autocorr_single, label='Señal con Ruido', alpha=0.7, color='#E41A1C', linestyle='--') # Rojo
    axes_autocorr[i].plot(lags_time, autocorr_avg, label='Señal Promediada', linewidth=2, color='#4DAF4A', marker='o', markersize=4, markevery=500) # Verde
    axes_autocorr[i].set_title(f'Frecuencia: {freq} Hz', fontsize=14)
    axes_autocorr[i].set_xlabel('Retardo (s)', fontsize=12)
    axes_autocorr[i].set_ylabel('Autocorrelación Normalizada', fontsize=12)
    axes_autocorr[i].set_xlim(-0.05, 0.05)
    axes_autocorr[i].legend(loc='upper right')

    # --- Histograma de Amplitud ---
    # Usar un rango de bins para que los histogramas sean comparables
    bins = np.linspace(-3, 3, 50)
    axes_hist[i].hist(signal_single_noisy, bins, density=True, alpha=0.7, label='Señal con Ruido', color='#E41A1C')
    axes_hist[i].hist(signal_averaged, bins, density=True, alpha=0.7, label='Señal Promediada', color='#4DAF4A')
    axes_hist[i].set_title(f'Distribución de Amplitud a {freq} Hz', fontsize=14)
    axes_hist[i].set_xlabel('Amplitud', fontsize=12)
    axes_hist[i].set_ylabel('Densidad de Probabilidad', fontsize=12)
    axes_hist[i].legend()

# --- 4. Finalizar y Mostrar Gráficos ---
fig_psd.tight_layout(rect=[0, 0, 1, 0.96])
fig_autocorr.tight_layout(rect=[0, 0, 1, 0.96])
fig_hist.tight_layout(rect=[0, 0, 1, 0.96])

# --- Imprimir tabla de resultados ---
print("\n--- Resultados de Media y Varianza ---")
print(f"{'Frecuencia':<12} | {'Media (Ruido)':<15} | {'Media (Prom.)':<15} | {'Varianza (Ruido)':<18} | {'Varianza (Prom.)':<18}")
print("-" * 85)
for i in range(len(frequencies)):
    print(f"{stats_results['freq'][i]:<12} | {stats_results['mean_single'][i]:<15.4f} | {stats_results['mean_avg'][i]:<15.4f} | {stats_results['var_single'][i]:<18.4f} | {stats_results['var_avg'][i]:<18.4f}")

# --- Crear gráfico de barras para la varianza ---
fig_stats, ax_stats = plt.subplots(figsize=(10, 6))
x = np.arange(len(frequencies))  # las localizaciones de las etiquetas
width = 0.35  # el ancho de las barras

# Colores cambiados a rojo y verde
rects1 = ax_stats.bar(x - width/2, stats_results['var_single'], width, label='Señal con Ruido', color='#E41A1C', edgecolor='black') # Rojo
rects2 = ax_stats.bar(x + width/2, stats_results['var_avg'], width, label='Señal Promediada', color='#4DAF4A', edgecolor='black') # Verde

# Añadir texto, títulos y etiquetas
ax_stats.set_ylabel('Varianza', fontsize=12)
ax_stats.set_title('Comparación de la Varianza por Frecuencia', fontsize=16)
ax_stats.set_xticks(x)
ax_stats.set_xticklabels([f'{f} Hz' for f in frequencies])
ax_stats.legend()
ax_stats.bar_label(rects1, padding=3, fmt='%.3f')
ax_stats.bar_label(rects2, padding=3, fmt='%.3f')
fig_stats.tight_layout()

plt.show()
```

```python
  import torch
import numpy as np
import matplotlib.pyplot as plt

# Establecer semilla para reproducibilidad
torch.manual_seed(42)

# Parámetros
T = 1000  # Número de puntos en el tiempo
Fs = 1000  # Frecuencia de muestreo
t = torch.arange(T) / Fs

N = 1000 # Número de realizaciones (promedios)
sigma = 0.5  # Desviación estándar del ruido

# Parámetros del armónico
A = 1.0  # Amplitud
f0 = 50.0  # Frecuencia en Hz
phi = torch.pi / 4 # Fase
```

```python
# Inicializar la suma de las señales
R_sum = torch.zeros(T)

# Bucle para promediar las realizaciones
for m in range(N):
    # Generar ruido gaussiano (n(t) en la notación)
    n = sigma * torch.randn(T)

    # Generar el armónico (s(t) en la notación)
    s = A * torch.cos(2 * torch.pi * f0 * t + phi)

    # Generar la señal ruidosa
    r = s + n

    # Sumar la señal ruidosa
    R_sum += r

# Calcular el promedio de las señales
R_average = R_sum / N
```

```python
# Calcular la FFT de la señal ruidosa (una sola realización)
fft_r = torch.fft.fft(r)
# Calcular la FFT de la señal promediada
fft_R_average = torch.fft.fft(R_average)

# Frecuencias correspondientes a la FFT
frequencies = torch.fft.fftfreq(T, d=1/Fs)

# Magnitudes
magnitude_fft_r = torch.abs(fft_r)
magnitude_fft_R_average = torch.abs(fft_R_average)
```

```python
# --- Gráfico en el Dominio del Tiempo ---
plt.figure(figsize=(14, 6))
plt.plot(t.numpy(), r.numpy(), label='Señal Ruidosa (una realización)', color='gray', alpha=0.6)
plt.plot(t.numpy(), R_average.numpy(), label=f'Señal Promediada (N={N})', color='green', linewidth=2)
plt.plot(t.numpy(), s.numpy(), '--', label='Armónico Puro (ideal)', color='red', alpha=0.7)
plt.title('Comparación en el Dominio del Tiempo')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.legend()
plt.grid(True)
plt.show()

# --- Gráfico en el Dominio de la Frecuencia ---
# Graficar solo las frecuencias positivas
positive_freqs_mask = frequencies >= 0

plt.figure(figsize=(14, 6))
plt.semilogy(frequencies[positive_freqs_mask].numpy(), magnitude_fft_r[positive_freqs_mask].numpy(), label='FFT de Señal Ruidosa')
plt.semilogy(frequencies[positive_freqs_mask].numpy(), magnitude_fft_R_average[positive_freqs_mask].numpy(), label='FFT de Señal Promediada', linewidth=2, color='green')
plt.title('Comparación del Espectro de Amplitud')
plt.xlabel('Frecuencia (Hz)')
plt.ylabel('Magnitud de la FFT')
plt.xlim(0, 200)
plt.ylim(1e-1, 2e3)
plt.legend()
plt.grid(True)
plt.show()
```
